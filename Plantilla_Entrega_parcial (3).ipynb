```python
# Escribe tu código aquí.
%pip install matplotlib
import math
import matplotlib.pyplot as plt

def ver_lista(heap):
   for tarea in heap:
       print(tarea)

def ver_niveles(heap):
   nivel = 0
   cantidad = 1
   i = 0
   while i < len(heap):
       print("Nivel", nivel, ": ", end="")
       cont = 0
       while cont < cantidad and i < len(heap):
           print(heap[i][0], "(", heap[i][2], ")", end="   ")
           i += 1
           cont += 1
       print()
       cantidad = cantidad * 2
       nivel += 1

def max_heapify(heap, n, i):
   mayor = i
   izq = 2 * i + 1
   der = 2 * i + 2
   if izq < n and heap[izq][2] > heap[mayor][2]:
       mayor = izq
   if der < n and heap[der][2] > heap[mayor][2]:
       mayor = der
   if mayor != i:
       aux = heap[i]
       heap[i] = heap[mayor]
       heap[mayor] = aux
       max_heapify(heap, n, mayor)

def build_max_heap(heap):
   n = len(heap)
   for i in range(n // 2 - 1, -1, -1):
       max_heapify(heap, n, i)

def extract_max(heap):
   maximo = heap[0]
   heap[0] = heap[len(heap) - 1]
   heap.pop()
   max_heapify(heap, len(heap), 0)
   return maximo

def agregar_tarea(heap, nueva):
   heap.append(nueva)
   hijo = len(heap) - 1
   while hijo > 0:
       padre = (hijo - 1) // 2
       if heap[hijo][2] > heap[padre][2]:
           aux = heap[hijo]
           heap[hijo] = heap[padre]
           heap[padre] = aux
           hijo = padre
       else:
           break

tareas = [
   ["T01", "Configurar servidor", 6],
   ["T02", "Reparar sensor", 10],
   ["T03", "Actualizar software", 4],
   ["T04", "Revisar camara", 8],
   ["T05", "Generar respaldo", 3],
   ["T06", "Configurar alarma", 7],
]
# punto 1
print("Arreglo inicial:")
ver_lista(tareas)
build_max_heap(tareas)
print("\nMax heap:")
ver_lista(tareas)
print("\nArbol:")
ver_niveles(tareas)
# punto 2
primera = extract_max(tareas)
print("\nSe proceso:", primera)
ver_lista(tareas)
segunda = extract_max(tareas)
print("\nSe proceso:", segunda)
ver_lista(tareas)
agregar_tarea(tareas, ["T07", "Falla en sistema contra incendios", 12])
print("\nDespues de insertar T07:")
ver_lista(tareas)
print("\nArbol final:")
ver_niveles(tareas)
# punto 3
nombres = ["A", "B", "C", "D", "E", "F"]
x = [2, 7, 4, 10, 13, 15]
y = [2, 3, 7, 8, 3, 7]
plt.scatter(x, y)
for i in range(6):
   plt.text(x[i], y[i], nombres[i])
plt.title("Dispositivos")
plt.grid()
plt.show()
aristas = []
for i in range(6):
   for j in range(i + 1, 6):
       d = math.sqrt((x[j] - x[i]) ** 2 + (y[j] - y[i]) ** 2)
       d = round(d, 2)
       aristas.append([i, j, d])
print("\nDistancias:")
for a in aristas:
   print(nombres[a[0]], "-", nombres[a[1]], ":", a[2])
# punto 4
for i in range(len(aristas)):
   for j in range(len(aristas) - 1):
       if aristas[j][2] > aristas[j + 1][2]:
           aux = aristas[j]
           aristas[j] = aristas[j + 1]
           aristas[j + 1] = aux
grupo = [0, 1, 2, 3, 4, 5]
mst = []
total = 0
for a in aristas:
   g1 = grupo[a[0]]
   g2 = grupo[a[1]]
   if g1 != g2:
       mst.append(a)
       total = total + a[2]
       for k in range(6):
           if grupo[k] == g2:
               grupo[k] = g1
   if len(mst) == 5:
       break
print("\nConexiones del MST:")
for a in mst:
   print(nombres[a[0]], "-", nombres[a[1]], ":", a[2])
total = round(total, 2)
print("Longitud total de cable:", total)
# punto 5
plt.scatter(x, y)
for i in range(6):
   plt.text(x[i], y[i], nombres[i])
for a in mst:
   plt.plot([x[a[0]], x[a[1]]], [y[a[0]], y[a[1]]])
   plt.text((x[a[0]] + x[a[1]]) / 2, (y[a[0]] + y[a[1]]) / 2, a[2])
plt.title("MST - Total de cable: " + str(total))
plt.grid()
plt.show()
```
