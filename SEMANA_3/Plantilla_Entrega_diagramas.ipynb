## 7) Análisis de complejidad **[15 pts]**
 1. Función T(n) y Justificación:

El tiempo total T(n) se obtiene sumando lo que tarda cada parte del programa:

T(n) = T_lectura(n) + T_MergeSort(n) + T_BusquedaBinaria(n)

Reemplazando por sus funciones de tiempo:

T(n) = c1 * n + c2 * (n * log2(n)) + c3 * log2(n) + c4

-c1 * n (Lectura): Pide e ingresa los n elementos uno a uno mediante un ciclo.
-c2 * (n * log2(n)) (Merge Sort): Divide la lista a la mitad recursivamente y la ordena al combinarla.
-c3 * log2(n) (Búsqueda Binaria): Reduce el espacio de búsqueda a la mitad en cada paso.
-c4 (Constantes): Operaciones simples de asignación e impresión.

Conclusión: La complejidad dominante del programa es O(n log n).

2. Mejor y Peor Caso
- Mejor Caso - Ω(n log n): Ocurre cuando el número buscado está justo en el centro de la lista. El tiempo total se mantiene en n log n.
- Peor Caso - O(n log n): Ocurre cuando el número no existe o está al final de la lista. El tiempo total sigue estando dominado por Merge Sort en n log n.

## 8) Código funcional **[15 pts]**



```python
# Escribe tu código aquí.
import math

def merge(A, p, q, r):
    L = A[p:q+1]
    R = A[q+1:r+1]

    i = 0
    j = 0
    for k in range(p, r + 1):
        if i < len(L) and (j >= len(R) or L[i] <= R[j]):
            A[k] = L[i]
            i += 1
        else:
            A[k] = R[j]
            j += 1

def merge_sort(A, p, r):
    if p < r:
        q = math.floor((p + r) / 2)
        merge_sort(A, p, q)
        merge_sort(A, q + 1, r)
        merge(A, p, q, r)

def binary_search(A, x):
    lo, hi = 0, len(A) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if A[mid] == x:
            return mid
        elif A[mid] < x:
            lo = mid + 1
        else:
            hi = mid - 1
    return None

# --- NUEVA FUNCIÓN ---
def obtener_entero(mensaje):
    """Pide un valor al usuario hasta que ingrese un número entero válido."""
    while True:
        try:
            return int(input(mensaje))
        except ValueError:
            print("Error: Entrada inválida. Por favor, ingrese solo números enteros.")
# ---------------------

def tienda_programa():
    # Nos aseguramos de que la cantidad sea un número mayor a 0
    while True:
        n = obtener_entero("Ingrese la cantidad de productos: ")
        if n > 0:
            break
        print("La cantidad debe ser mayor a 0.")

    codigos = []
    for i in range(n):
        # Usamos nuestra nueva función en lugar de int(input(...))
        codigo = obtener_entero(f"Ingrese el código del producto {i+1}: ")
        codigos.append(codigo)

    merge_sort(codigos, 0, len(codigos) - 1)
    print(f"\nLista de códigos ordenada: {codigos}")

    # También la usamos para la búsqueda
    objetivo = obtener_entero("\nIngrese el código que desea buscar: ")
    resultado = binary_search(codigos, objetivo)

    if resultado is not None:
        print(f"Mensaje: El producto con código {objetivo} EXISTE (encontrado en posición {resultado}).")
    else:
        print(f"Mensaje: El producto con código {objetivo} NO existe en el programa.")

if __name__ == "__main__":
    tienda_programa()
```
