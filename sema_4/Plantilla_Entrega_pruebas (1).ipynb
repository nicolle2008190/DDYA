## 7) Análisis de complejidad **[5 pts]**
Justificación: El programa usa un ciclo para el valor monto (n=cambio/50) y un ciclo interno fijo de 5 monedas. La cantidad de monedas al ser inavriante hace que las operaciones crecezcan de forma directamente proporcional al tamaño del cambio.

Función T(n): T(n)=5n+c⟹O(n) (Lineal).
Complejidad Espacial: O(n) por la lista dp de tamaño n+1.

Mejor y Peor Caso:
Mejor Caso O(1): Monto inválido o no múltiplo de 50 .
Peor Caso O(n): Monto válido donde se recorre la lista completa desde 1 hasta n. La gráfica muestra una línea recta ascendente.


## 8) Código funcional **[10 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def calcular_cambio_dp(valor):

    # 1. Validaciones iniciales
    if valor <= 0 or valor % 50 != 0:
        return None, -1

    monedas = [50, 100, 200, 500, 1000]

    dp = [float('inf')] * (valor + 1)
    dp[0] = 0
    moneda_usada = [0] * (valor + 1)

    for i in range(50, valor + 1, 50):  # Avanzamos de 50 en 50
        for m in monedas:
            if i >= m and dp[i - m] + 1 < dp[i]:
                dp[i] = dp[i - m] + 1
                moneda_usada[i] = m

    if dp[valor] == float('inf'):
        return None, -1

    actual = valor
    desglose = []
    while actual > 0:
        m = moneda_usada[actual]
        desglose.append(str(m))
        actual -= m

    desglose.sort(key=lambda x: int(x), reverse=True)

    texto_suma = " + ".join(desglose)
    return texto_suma, dp[valor]

print("--- Sistema de Cambio (Programación Dinámica) ---")
entrada_usuario = input("Ingrese el valor del cambio deseado: ")

try:
    valor_ingresado = int(entrada_usuario)
    desglose, cantidad = calcular_cambio_dp(valor_ingresado)

    if cantidad == -1:
        print("Error: El valor debe ser mayor a cero y múltiplo de 50.")
    else:
        print(f"\n{desglose} = {valor_ingresado}")
        print(f"Cantidad mínima de monedas: {cantidad}")

except ValueError:
    print("Error: Por favor ingrese un número entero válido.")
```

#### Código base para benchmarking

```python
def bench(n: int, step: int = 50) -> list[tuple[int, float]]:
    print("Generando datos ...", end='')
    import time
    results: list[tuple[int, float]] = []

    # Evalúa el impacto desde 50 pesos hasta el tope (n)
    for size in range(50, n, step):
        ti = time.perf_counter()
        output = calcular_cambio(size)
        tf = time.perf_counter()

        delta = (tf - ti) * 1000 # Convertir a milisegundos
        results.append((size, delta))
    print(" Listo.")
    return results

def dibujar(x, y, dimensiones=(7, 5)):
    assert len(x) == len(y)
    import matplotlib.pyplot as plt
    plt.figure(figsize=dimensiones)
    plt.plot(x, y, marker='.', linestyle='none', color='#2ca02c')
    plt.title('Complejidad Empírica: Cambio de Monedas')
    plt.xlabel('Valor del Cambio en pesos (V)')
    plt.ylabel('Tiempo (milisegundos)')
    plt.show()

# Probamos con cambios desde 50 pesos hasta 10,000 pesos
tiempos = bench(10000, step=100)

x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]

dibujar(x, y)
```
