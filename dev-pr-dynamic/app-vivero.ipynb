La recursión simple para Fibonacci tiene un tiempo exponencial 0(2^n) porque repite los mismos subproblemas múltiples veces. Al usar memoización, cada subproblema desde 0 hasta n se calcula una sola vez y se almacena en memoria.

 T(n)=0(n)
- Mejor y peor caso con gráfica

Mejor caso — O(1): Ocurre cuando n=0, n=1, o cuando el valor de n ya está previamente guardado en el diccionario de memoización.

Peor caso — O(n): Ocurre la primera vez que se consulta un mes n no calculado, requiriendo llenar la tabla de memoria desde 0 hasta n.

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
# Escribe tu código aquí.
memo = {0: 0, 1: 1}
def calcular_plantas(n):
    if n in memo:
        return memo[n]
    memo[n] = calcular_plantas(n - 1) + calcular_plantas(n - 2)
    return memo[n]
while True:
    try:
        mes = int(input("Ingrese el mes: "))
        if mes < 0:
            print("Por favor, ingrese un número igual o mayor a 0.")
            continue
        resultado = calcular_plantas(mes)
        print(f"La producción del mes {mes} es: {resultado}")
        break
    except ValueError:
        print("Error: Ingreso inválido. Por favor, escriba únicamente números enteros.")
```
